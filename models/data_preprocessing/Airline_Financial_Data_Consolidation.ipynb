```python
import pandas as pd
from pathlib import Path
import re

BASE_PATH = Path("/Users/sara.alsiyat/Desktop/Uni/Study/Winter/420 - Database systems/Final Project")

FIN_DIR = BASE_PATH / "data" / "03_Financial"
OUT_DIR = FIN_DIR / "merged"
OUT_DIR.mkdir(parents=True, exist_ok=True)

FILES = [
    FIN_DIR / "Airline_Financial_2023.csv",
    FIN_DIR / "Airline_Financial_2024.csv",
    FIN_DIR / "Airline_Financial_2025.csv",
]

OUTPUT_FILE = OUT_DIR / "Airline_Financial_2023_2025_ALL.csv"


# MERGE

dfs = []

for f in FILES:
    if not f.exists():
        raise FileNotFoundError(f"File not found: {f}")

    df = pd.read_csv(f, low_memory=False)

    # Add Year if missing
    if "Year" not in df.columns:
        m = re.search(r"(20\d{2})", f.name)
        df["Year"] = int(m.group(1)) if m else None

    # Keep track of source file 
    df["source_file"] = f.name

    dfs.append(df)

# Concat (handles different columns automatically)
merged = pd.concat(dfs, ignore_index=True, sort=False)

# put Year first (nice formatting)
cols = list(merged.columns)
if "Year" in cols:
    cols.insert(0, cols.pop(cols.index("Year")))
merged = merged[cols]

# Save
merged.to_csv(OUTPUT_FILE, index=False)

print("✅ Merged file saved to:")
print(OUTPUT_FILE)
print("\nRows:", len(merged))
print("Columns:", len(merged.columns))
print("Year counts:\n", merged["Year"].value_counts(dropna=False).sort_index())
```
