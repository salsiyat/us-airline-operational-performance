```python
import pandas as pd
from pathlib import Path
import glob

BASE_PATH = Path(
    "/Users/sara.alsiyat/Desktop/Uni/Study/Winter/420 - Database systems/Final Project/us-airline-operational-performance-main"
)

ON_TIME_PATH = BASE_PATH / "data" / "01_OnTime_Performance"
OUTPUT_PATH = ON_TIME_PATH / "merged"

OUTPUT_PATH.mkdir(parents=True, exist_ok=True)

YEARS = [2023, 2024, 2025]


# MERGE FUNCTION

def merge_year(year: int):
    year_dir = ON_TIME_PATH / str(year)
    files = sorted(glob.glob(str(year_dir / f"OnTime_{year}_*.csv")))

    if not files:
        print(f"❌ No files found for {year}")
        return None

    output_file = OUTPUT_PATH / f"OnTime_{year}_ALL.csv"

    print(f"\n📦 Merging {len(files)} files for {year}")
    for f in files:
        print(" -", Path(f).name)

    first = True
    with open(output_file, "w", encoding="utf-8", newline="") as out:
        for i, file in enumerate(files, start=1):
            print(f"   [{i}/{len(files)}] Processing {Path(file).name}")
            df = pd.read_csv(file, low_memory=False)
            df.to_csv(out, index=False, header=first)
            first = False

    print(f"✅ Created: {output_file.name}")
    return output_file


# RUN PER-YEAR MERGES

merged_files = []

for year in YEARS:
    result = merge_year(year)
    if result:
        merged_files.append(result)

# MERGE ALL YEARS

if merged_files:
    combined_file = OUTPUT_PATH / "OnTime_2023_2025_ALL.csv"
    print("\n📊 Creating combined file for all years")

    first = True
    with open(combined_file, "w", encoding="utf-8", newline="") as out:
        for f in merged_files:
            for chunk in pd.read_csv(f, chunksize=300_000, low_memory=False):
                chunk.to_csv(out, index=False, header=first)
                first = False

    print(f"🎉 Combined file created: {combined_file.name}")

print("\nDONE ✅")
print("Merged files are located in:")
print(OUTPUT_PATH)
```
